# Small Cap Excel to CSV Conversion

Converts only the five fund-specific Excel workbooks, using each workbook’s `Historical NAV` sheet. This creates exactly five CSV inputs in `Small_cap/clean`, matching the five fund files in the large-cap workflow. The two combined dataset workbooks are reference data, not fund inputs.


```python
from pathlib import Path
import pandas as pd
```

```python
funds = {
    "Nippon India Small Cap": ("Small_cap/Nippon_India_Small_Cap_Fund_2022_2025.xlsx", "nippon_small_cap.csv"),
    "SBI Small Cap": ("Small_cap/SBI_Small_Cap_Fund_2022_2025.xlsx", "sbi_small_cap.csv"),
    "HDFC Small Cap": ("Small_cap/HDFC_Small_Cap_Fund_2022_2025.xlsx", "hdfc_small_cap.csv"),
    "Axis Small Cap": ("Small_cap/Axis_Small_Cap_Fund_2022_2025.xlsx", "axis_small_cap.csv"),
    "Kotak Small Cap": ("Small_cap/Kotak_Small_Cap_Fund_2022_2025.xlsx", "kotak_small_cap.csv")
}

output_dir = Path("Small_cap/clean")
output_dir.mkdir(parents=True, exist_ok=True)

for fund_name, (file_path, output_name) in funds.items():
    output_file = output_dir / output_name
    nav_data = pd.read_excel(file_path, sheet_name="Historical NAV")
    nav_data.to_csv(output_file, index=False)
    print(f"Saved: {output_file}")
```
