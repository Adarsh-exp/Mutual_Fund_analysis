# Mid Cap Excel to CSV Conversion

Exports the `Historical NAV Data` sheet from each of the five updated mid-cap Excel workbooks to one CSV per fund in `Mid_cap/csv`.

```python
from pathlib import Path
import pandas as pd
```

```python
funds = {
    "Aditya Birla Sun Life Midcap Fund": ("Mid_cap/aditya_mid_cap.xlsx", "aditya_mid_cap.csv"),
    "Axis Midcap Fund": ("Mid_cap/axis_mid_cap.xlsx", "axis_mid_cap.csv"),
    "DSP Midcap Fund": ("Mid_cap/dsp_mid_cap.xlsx", "dsp_mid_cap.csv"),
    "HDFC Mid Cap Fund": ("Mid_cap/hdfc_mid_cap.xlsx", "hdfc_mid_cap.csv"),
    "Mahindra Manulife Mid Cap Fund": ("Mid_cap/mahindra_mid_cap.xlsx", "mahindra_mid_cap.csv")
}

output_dir = Path("Mid_cap/csv")
output_dir.mkdir(parents=True, exist_ok=True)

for fund_name, (file_path, output_name) in funds.items():
    nav_data = pd.read_excel(file_path, sheet_name="Historical NAV Data", header=4)
    nav_data.columns = nav_data.columns.astype(str).str.strip()
    output_path = output_dir / output_name
    nav_data.to_csv(output_path, index=False)
    print(f"Saved {fund_name}: {output_path} ({len(nav_data)} rows)")
```
